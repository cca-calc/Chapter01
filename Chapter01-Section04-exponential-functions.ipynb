<div align="center">

| Law | Rule | Example |
|---|---|---|
| Product of powers | $a^m \cdot a^n = a^{m+n}$ | $x^2 \cdot x^3 = x^5$ |
| Quotient of powers | $\dfrac{a^m}{a^n} = a^{m-n}$ | $\dfrac{x^5}{x^2} = x^3$ |
| Power of a power | $(a^m)^n = a^{mn}$ | $(x^2)^3 = x^6$ |
| Power of a product | $(ab)^n = a^n b^n$ | $(2x)^3 = 8x^3$ |
| Power of a quotient | $\left(\dfrac{a}{b}\right)^n = \dfrac{a^n}{b^n}$ | $\left(\dfrac{x}{2}\right)^3 = \dfrac{x^3}{8}$ |
| Zero exponent | $a^0 = 1$ | $x^0 = 1$ |
| Negative exponent | $a^{-n} = \dfrac{1}{a^n}$ | $x^{-3} = \dfrac{1}{x^3}$ |
| Fractional exponent | $a^{1/n} = \sqrt[n]{a}$ | $x^{1/2} = \sqrt{x}$ |
| General fractional exponent | $a^{m/n} = \sqrt[n]{a^m}$ | $x^{2/3} = \sqrt[3]{x^2}$ 

</div>

**Conditions:** $a \ne 0$ and $b \ne 0$ where they appear in a denominator.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = 1
npoints = 99
x = np.linspace(-a, a, npoints)
```

#### Problem #5

```python
y0 = (2)**(x)
y1 = (np.e)**(x)
y2 = (5)**(x)
y3 = (20)**(x) 
```

```python
plt.plot(x, y0, label=r'$2^x$', color='black')
plt.plot(x, y1, label=r'$e^x$', color='red')
plt.plot(x, y2, label=r'$5^x$', color='green')
plt.plot(x, y3, label=r'$-20^x$', color='blue')
plt.title('Exponential function with different base.')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
a = 3
npoints = 99
xpos = np.linspace(0.05, a, npoints)

xneg = np.linspace(-a, -0.05, npoints)
fxp0 = (1 - np.e ** (1 / xpos)) / (1 + np.e ** (1/xpos))
fxn0 = (1 - np.e ** (1 / xneg)) / (1 + np.e ** (1/xneg))
fxp1 = (1 - np.e ** (-1 / xpos)) / (1 + np.e ** (-1/xpos))
fxn1 = (1 - np.e ** (-1 / xneg)) / (1 + np.e ** (-1/xneg))
plt.plot(xpos, fxp0, color='black')
plt.plot(xneg, fxn0, color='black')
plt.plot(xpos, fxp1, color='red')
plt.plot(xneg, fxn1, color='red')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```
