### Combination and Composite of Functions

<div align="center">

| Operation | Equation |
|---|---|
| Sum | $(f+g)(x) = f(x) + g(x)$ |
| Difference | $(f-g)(x) = f(x) - g(x)$ |
| Product | $(f\cdot g)(x) = f(x)\cdot g(x)$ |
| Division | $\left(\frac{f}{g}\right)(x) = \frac{f(x)}{g(x)}, \quad g(x)\neq 0$ |
| Composite | $(f\circ g)(x) = f(g(x))$ |

</div>
