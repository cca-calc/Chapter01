## Power Functions
- For power functions, The domain is $(-\infty, \infty)$ 
- Set the Domain: the x values from (-a to a) and a with n values

```python
a = 1
npoints = 100
x = np.linspace(-a, a, npoints)
y4 = x**4
y5 = x**5
```

```python
fx = x**2
plt.plot(x, fx, color='red')
plt.title('$f(x)= x^{2}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python
fx = x**3
plt.plot(x, fx, color='red')
plt.title('$f(x)= x^{3}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python
fx = x**4
plt.plot(x, fx, color='red')
plt.title('$f(x)= x^{4}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python
fx = x**5
plt.plot(x, fx, color='red')
plt.title('$f(x)= x^{5}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

## Root Functions
- For root functions, The domain depends on if the root is even or odd 
- For even roots (2,4,...) it is $(0, \infty)$  
- For odd roots (3,5,...) it is $(-\infty, \infty)$

```python
a = 1
npoints = 100
xeven = np.linspace(0.01, a, npoints)
xodd = np.linspace(-a, a, npoints)
```

```python

fx = np.sqrt(xeven)
plt.plot(xeven, fx, color='red')
plt.title('$f(x)= x^{1/2}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python

fx = np.cbrt(xodd)
plt.plot(xodd, fx, color='red')
plt.title('$f(x)= x^{1/3}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python

fx = xeven**(1/4)
plt.plot(xeven, fx, color='red')
plt.title('$f(x)= x^{1/4}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python

fx = np.sign(xodd) * np.abs(xodd) ** (1/5)
plt.plot(xodd, fx, color='red')
plt.title('$f(x)= x^{1/5}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

## Reciprocal Functions
- For reciprocal functions, The domain is $(-\infty, 0)$ U $(0, \infty)$ 
- Set the Domain: the x values from (-a to -0.01) and a with n values

```python
a = 0.25
npoints = 100
xneg = np.linspace(-a, -0.01, npoints)
xpos = np.linspace(0.01, a, npoints)
```

- Plot the functions on the domain [-a,a]
- Add a title, labels, grid, labels, and legend

##### Reciprocal Functions change rapidly with power. Replot these individually and see that range change.

```python

fx = xneg**(-1)
plt.plot(xneg, fx, color='red')
fx = xpos**(-1)
plt.plot(xpos, fx, color='green')
plt.title('$f(x)= x^{-1}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python
fx = xneg**(-2)
plt.plot(xneg, fx, color='red')
fx = xpos**(-2)
plt.plot(xpos, fx, color='green')
plt.title('$f(x)= x^{-2}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```

```python
fx = xneg**(-3)
plt.plot(xneg, fx, color='red')
fx = xpos**(-3)
plt.plot(xpos, fx, color='green')
plt.title('$f(x)= x^{-3}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.show()
```
