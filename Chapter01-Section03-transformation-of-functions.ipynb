### Chapter 1, Section 3: The Tranformation of functions: shifting, streching, reflections.
- Problems: Paper (35, 37, 39, 45 and 47).

```python
import numpy as np
import matplotlib.pyplot as plt
```

- Set the Domain: the x values from -a to -a with npoint values

```python
a = 3
npoints = 200
x = np.linspace(-a, a, npoints)
```

```python
c = 3
n = 1
```

- Plot the functions on the domain [-a,a]
- Add a title, labels, grid, labels, and legend

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = x**(n) + c #shift up
plt.plot(x, fx, label=r'$x^n + c$', color='red')
fx = x**(n) - c #shift down
plt.plot(x, fx, label=r'$x^n - c$', color='green')
plt.title('Vertical Shifting (up/down)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = (x + c)**(n) #shift left 
plt.plot(x, fx, label=r'$(x+c)^n$', color='red')
fx = (x - c)**(n) #shift right
plt.plot(x, fx, label=r'$(x-c)^n$', color='green')
y3 = (x + c)**(n) 
plt.title('Horizontal Shifting (right/left)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = c*(x**(n)) #stretch vertically
plt.plot(x, fx, label=r'$c*x^n$', color='red')
fx = (1/c)*(x**(n)) #shrink vertically
plt.plot(x, fx, label=r'$(\frac{1}{c})*x^n$', color='green')
plt.title('Vertical Stretching and Shrinking')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python

fx = c*(x**(n)) 
plt.plot(x, fx, label=r'$x^n$', color='black')
fx =(c*x)**(n) #shrink horizontally
plt.plot(x, fx, label=r'$(c*x)^n$', color='red')
fx = (x/c)**(n) #stretch horizontally 
plt.plot(x, fx, label=r'$(\frac{1}{c}*x)^n$', color='green')
plt.title('Horizontal Stretching and Shrinking')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx =-(x**n)  #reflect about x axis (shifted down by 0.5 to make the line visible). 
plt.plot(x, fx, label=r'$-(x)^n$', color='red')
fx = (-x)**(n) #reflect about the y axis
plt.plot(x, fx, label=r'$(-x)^n$', color='green')
plt.title('Reflecting')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```
