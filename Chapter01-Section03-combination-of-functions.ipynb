- This example below shows the various transformations of a power function.
- Feel free to change a, c, or n.

```python
import numpy as np
import matplotlib.pyplot as plt
```

- Set the Domain: the x values from -a to -a with npoint values

```python
a = 3
npoints = 200
x = np.linspace(-a, a, npoints)
```

```python
#print(x) #Remove the first hashtag to see the values of x.
```

- Define the functions: example below is for power functions
- Do the same for the vertical, horizontal, stretching and reflection as shown on page 37.

```python
c = 3
n = 3
```

```python
#print(y1) #Remove the first hashtag to see the values of y1.
#print(y4) #Remove the first hashtag to see the values of y2.
```

- Plot the functions on the domain [-a,a]
- Add a title, labels, grid, labels, and legend

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = x**(n) + c
plt.plot(x, fx, label=r'$x^n + c$', color='red')
fx = x**(n) - c
plt.plot(x, fx, label=r'$x^n - c$', color='green')
plt.title('Vertical Shifting')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = (x + c)**(n)
plt.plot(x, fx, label=r'$(x+c)^n$', color='red')
fx = (x - c)**(n)
plt.plot(x, fx, label=r'$(x-c)^n$', color='green')
y3 = (x + c)**(n) 
plt.title('Horizontal Shifting')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = c*(x**(n))
plt.plot(x, fx, label=r'$(c*x)^n$', color='red')
fx = (1/c)*(x**(n))
plt.plot(x, fx, label=r'$(\frac{1}{c})*x^n$', color='green')
plt.title('Vertical Stretching and Shrinking')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python

fx = c*(x**(n))
plt.plot(x, fx, label=r'$x^n$', color='black')
fx =(c*x)**(n)
plt.plot(x, fx, label=r'$(c*x)^n$', color='red')
fx = (x/c)**(n)
plt.plot(x, fx, label=r'$(\frac{1}{c}*x)^n$', color='green')
plt.title('Horizontal Stretching and Shrinking')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx =-(x**(n)) - 2  #shifted by 2 to make the line visible. 
plt.plot(x, fx, label=r'$-(x)^n$', color='red')
fx = (-x)**(n)
plt.plot(x, fx, label=r'$(-x)^n$', color='green')
plt.title('Reflecting')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```
