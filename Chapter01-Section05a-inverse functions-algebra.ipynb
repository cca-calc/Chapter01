```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_vals = np.linspace(0, 2, 400)
y_vals = 1 - x_vals**2

y_vals_inv = np.linspace(1, -3, 400)
x_vals_inv = np.sqrt(1 - y_vals_inv)

plt.plot(x_vals, y_vals, label='$f(x) = 1 - x^2$', color='red')
plt.plot(y_vals_inv, x_vals_inv, label='$f(x) = (1-x)^{0.5}$', color='green')
plt.plot(x_vals, x_vals, color='black', linestyle=':', label='y = x')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Function and Its Inverse')
plt.legend()
plt.grid(True)
plt.axis('equal')
plt.show()
```

```python
x_vals = np.linspace(-1, 8, 100)
y_vals = 2 + np.sqrt(x_vals + 1)

y_vals_inv = np.linspace(2, 5, 100)
x_vals_inv = (y_vals_inv - 2)**2 - 1

plt.plot(x_vals, y_vals, label='$f(x) = 2 + (x+1)^{0.5}$', color='red')
plt.plot(y_vals_inv, x_vals_inv, label='$f(x)=(x-2)^2 - 1$', color='green')
plt.plot(x_vals, x_vals, color='black', linestyle=':', label='y = x')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Function and Its Inverse')
plt.legend()
plt.grid(True)
plt.axis('equal')
plt.show()
```
