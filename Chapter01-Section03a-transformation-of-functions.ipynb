### Chapter 1, Section 3: The Transformation of functions: shifting, streching, reflections.
- Problems: Jupyter (1). (plotted below)

```python
import numpy as np
import matplotlib.pyplot as plt
```

- Set the Domain: the x values from -a to -a with npoint values

```python
a = 5
npoints = 200
x = np.linspace(-a, a, npoints)
```

```python
c = 5
n = 3
```

- Plot the functions on the domain [-a,a]
- Add a title, labels, grid, labels, and legend

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = x**(n) + c #shift up
plt.plot(x, fx, label=r'$x^n + c$', color='red')
fx = x**(n) - c #shift down
plt.plot(x, fx, label=r'$x^n - c$', color='green')
plt.title('Vertical Shifting (up/down)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = (x + c)**(n) #shift left 
plt.plot(x, fx, label=r'$(x+c)^n$', color='red')
fx = (x - c)**(n) #shift right
plt.plot(x, fx, label=r'$(x-c)^n$', color='green')
y3 = (x + c)**(n) 
plt.title('Horizontal Shifting (right/left)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n)
plt.plot(x, fx, label=r'$x^n$', color='black')
fx = c*(x**(n)) #stretch vertically
plt.plot(x, fx, label=r'$c*x^n$', color='red')
fx = (1/c)*(x**(n)) #shrink vertically
plt.plot(x, fx, label=r'$(\frac{1}{c})*x^n$', color='green')
plt.title('Vertical Stretching and Shrinking')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python

fx = c*(x**(n)) 
plt.plot(x, fx, label=r'$x^n$', color='black')
fx =(c*x)**(n) #shrink horizontally
plt.plot(x, fx, label=r'$(c*x)^n$', color='red')
fx = (x/c)**(n) #stretch horizontally 
plt.plot(x, fx, label=r'$(\frac{1}{c}*x)^n$', color='green')
plt.title('Horizontal Stretching and Shrinking')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
fx = x**(n) + a
plt.plot(x, fx, label=r'$x^n +a$ ', color='black')
fx =-(x**n) - a   #reflect about x axis  
plt.plot(x, fx, label=r'$-(x)^n$ -a', color='red')
fx = (-x)**(n) + a #reflect about the y axis
plt.plot(x, fx, label=r'$(-x)^n +a$', color='green')
plt.title('Reflecting')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

### Function Transformations Summary

| New function | Transformation | What it does |
|---|---|---|
| $f(x)+c,\ c>0$ | Vertical shift up | Translates the graph upward by $c$ units |
| $f(x)-c,\ c>0$  | Vertical shift down | Translates the graph downward by $c$ units |
| $f(x+c),\ c>0$  | Horizontal shift left | Translates the graph left by $c$ units |
| $f(x-c),\ c>0$  | Horizontal shift right | Translates the graph right by $c$ units |
| $c\,f(x),\ c>1$ | Vertical stretch | Makes the graph taller by multiplying all $y$-values by $c$ |
| $\frac{1}{c}\,f(x),\ c>1$ | Vertical shrink | Makes the graph shorter by compressing the $y$-values |
| $f(c\,x),\ c>1$ | Horizontal shrink | Compresses the graph toward the $y$-axis |
| $f\!\left(\frac{x}{c}\right),\ c>1$ | Horizontal stretch | Stretches the graph away from the $y$-axis |
| $-f(x)$ | Reflection across the $x$-axis | Flips the graph upside down |
| $f(-x)$ | Reflection across the $y$-axis | Flips the graph left-to-right |
